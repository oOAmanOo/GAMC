```python
import pandas as pd
import json
```

Recreate image caption pairs dataset in jsonl format >> mcdonalds_switzerland_img

```python
data = pd.read_csv('../../Main/Model/3000/ablation/MC/_woFCT_onlyLoRaadapt/generate2_1.csv')
# data = pd.read_csv('../../Main/Model/clip/clip_100up_only200_lessNotFunImg_53_171_passlength_12_MC_with_oxford_3000_only1_300_82/generate2_1.csv')
# data = pd.read_csv('../../Main/Model/bita/mc/generate2_1.csv')

jsonFile = []
for i in range(data.shape[0]):
    caption = data['caption'][i].replace('\"', '\'').replace('\n', '').replace('\\', '').replace('\r', '')
    jsonFile.append(f'\u007b"id": "{data["image_id"][i]}", "img": "mcdonalds_switzerland_img/{data["image_id"][i]}.jpg", "text": "{caption}"\u007d')

path = './data/mine_53_171_MC_woFCT_onlyLoRaadapt_generate2.jsonl'
# path = './data/mine_53_171_MC_onlybitfitadapt_generate2.jsonl'
# path = './data/clip_MC_all_generate2_1.jsonl'
# path = 'data/bita_mc_all.jsonl'

with open(path, 'w', encoding='utf-8') as f:
    for item in jsonFile:
        f.write("%s\n" % item)
```

Recreate image caption pairs dataset in jsonl format >> sonicdrivein_img

```python
data = pd.read_csv('../../Main/Model/3000/ablation/SD/_woFCT_onlyLoRaadapt/generate2_1.csv')
# data = pd.read_csv('../../Main/Model/clip/clip_100up_only200_lessNotFunImg_169_55_passlength_10_SD_with_oxford_3000_only1_300_82/generate2_1.csv')
# data = pd.read_csv('../../Main/Model/bita/sd/generate2_1.csv')

jsonFile = []
for i in range(data.shape[0]):
    caption = data['caption'][i].replace('\"', '\'').replace('\n', '').replace('\\', '').replace('\r', '')
    jsonFile.append(f'\u007b"id": "{data["image_id"][i]}", "img": "sonicdrivein_img/{data["image_id"][i]}.jpg", "text": "{caption}"\u007d')

path = './data/mine_169_55_SD_woFCT_onlyLoRaadapt_generate2.jsonl'
# path = './data/mine_169_55_SD_onlybitfitadapt_generate2.jsonl'
# path = './data/clip_SD_all_generate2_1.jsonl'
# path = 'data/bita_sd_all.jsonl'

with open(path, 'w', encoding='utf-8') as f:
    for item in jsonFile:
        f.write("%s\n" % item)
```

Test if jsonl file is correct

```python
pre = ''
raw_data = []
with open(path, 'r', encoding='utf-8')as f:
    for line in f:
        if '\u007d' not in line:
            pre += line
            continue
        if pre:
            line = pre + line
            pre = ''
        print(line)
        data = json.loads(line)
        raw_data.append(data)
        print("=====")
print("Load %d data from split(s)." % (len(raw_data)))
```

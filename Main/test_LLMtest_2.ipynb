```python
from torch.nn.functional import cosine_similarity
image_G = torch.randn(32,64,20)
embedding_matrix = torch.randn(256000,20)
```

```python
embedding_matrix = embedding_matrix.unsqueeze(0).expand(image_G.shape[0], -1, -1)
print(image_G.shape,embedding_matrix.shape)
```

```python
print(image_G.unsqueeze(2).shape, embedding_matrix.unsqueeze(1).shape)
```

```python
import faiss
embedding_dim = 20
flattened_image_G = image_G.view(-1, embedding_dim)  # (batch_size * 32, embedding_dim)
# 初始化 FAISS 索引
index = faiss.IndexFlatL2(embedding_dim)  # 使用 L2 距離
embedding_matrix_np = embedding_matrix.cpu().numpy()  # 需要轉換為 NumPy
index.add(embedding_matrix_np)  # 加入向量到索引中

# 搜索最近鄰
flattened_image_G_np = flattened_image_G.cpu().numpy()  # (batch_size * 32, embedding_dim)
_, token_ids = index.search(flattened_image_G_np, 1)  # 搜索每個向量最近的 1 個鄰居
token_ids = torch.tensor(token_ids.squeeze(), device=image_G.device).view(image_G.size(0), image_G.size(1))
token_ids
```

```python
chunk_size = 256
token_ids = []
for i in range(0, embedding_matrix.size(1), chunk_size):
    chunk = embedding_matrix[:,i:i + chunk_size]  # (chunk_size, embedding_dim)
    print(image_G.unsqueeze(2).shape,chunk.shape)
    similarity_chunk = cosine_similarity(image_G.unsqueeze(2), chunk.unsqueeze(1), dim=-1)
    # similarity_chunk = cosine_similarity(flattened_image_G.unsqueeze(1), chunk.unsqueeze(0), dim=-1)  # (batch_size * 32, chunk_size)
    token_ids_chunk = similarity_chunk.argmax(dim=-1) + i  # Offset token IDs by chunk start index
    token_ids.append(token_ids_chunk)
print(token_ids)
```
